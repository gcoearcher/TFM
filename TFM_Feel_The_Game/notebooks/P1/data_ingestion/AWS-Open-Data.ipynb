```python
import boto3
import pandas as pd
from io import BytesIO

def read_aws_public_csv(bucket_name, object_key):
    s3 = boto3.client('s3')
    obj = s3.get_object(Bucket=bucket_name, Key=object_key)
    df = pd.read_csv(BytesIO(obj['Body'].read()))
    return df

# Example: NOAA Global Summary data (public)
bucket_name = "noaa-ghcn-pds"
object_key = "csv.gz/2020.csv.gz"

df = read_aws_public_csv(bucket_name, object_key)
print(df.head())
```
