```python
from delta import *
import pyspark

builder = pyspark.sql.SparkSession.builder.appName("Feel the game") \
    .config("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension") \
    .config("spark.sql.catalog.spark_catalog", "org.apache.spark.sql.delta.catalog.DeltaCatalog")
    
spark = configure_spark_with_delta_pip(builder).getOrCreate()
```

```python
df = spark.read.format("delta").load("../datasets/reviews.json")
df.show()
```

```python
# Load JSON (if needed)
df = spark.read.json("../datasets/reviews.json")

# Convert to Delta (Parquet-based)
df.write.format("delta").save("/path/to/delta_table")
```
